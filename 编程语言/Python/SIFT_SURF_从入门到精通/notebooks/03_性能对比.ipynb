# 03 SIFT、ORB、AKAZE 性能对比
这里的单次样例只用于理解指标。正式报告应预热、多次测量并加入更多图像。

```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'scripts'))
import cv2 as cv
from common import DEFAULT_IMAGE_A, DEFAULT_IMAGE_B, read_image, detect_and_match, metrics_dict
a = cv.cvtColor(read_image(DEFAULT_IMAGE_A), cv.COLOR_BGR2GRAY)
b = cv.cvtColor(read_image(DEFAULT_IMAGE_B), cv.COLOR_BGR2GRAY)
rows = []
for name in ['sift', 'orb', 'akaze']:
    result = detect_and_match(a, b, name, 0.75)
    rows.append(metrics_dict(result, name, 0.75))
rows
```

```python
assert all(row['good_matches'] > 0 for row in rows)
print([(row['algorithm'], row['good_matches'], row['inlier_ratio']) for row in rows])
```
