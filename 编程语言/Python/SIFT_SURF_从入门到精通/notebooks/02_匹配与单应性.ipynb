```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'scripts'))
import cv2 as cv
from common import DEFAULT_IMAGE_A, DEFAULT_IMAGE_B, read_image, detect_and_match, draw_matches, metrics_dict, write_image
image_a, image_b = read_image(DEFAULT_IMAGE_A), read_image(DEFAULT_IMAGE_B)
result = detect_and_match(cv.cvtColor(image_a, cv.COLOR_BGR2GRAY), cv.cvtColor(image_b, cv.COLOR_BGR2GRAY), 'sift', 0.75)
metrics = metrics_dict(result, 'sift', 0.75)
metrics
```
