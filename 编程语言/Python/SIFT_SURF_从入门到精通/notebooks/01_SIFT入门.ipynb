# 01 SIFT 入门
检测关键点、查看描述子形状，并保存带尺度与方向的关键点图。

```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'scripts'))
import cv2 as cv
from common import DEFAULT_IMAGE_A, read_image, draw_keypoints, write_image
image = read_image(DEFAULT_IMAGE_A)
gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
sift = cv.SIFT_create()
keypoints, descriptors = sift.detectAndCompute(gray, None)
print('关键点:', len(keypoints), '描述子:', descriptors.shape, descriptors.dtype)
```

```python
output = ROOT / 'outputs' / 'notebook_01_sift_keypoints.png'
write_image(output, draw_keypoints(image, keypoints))
print(output)
```
